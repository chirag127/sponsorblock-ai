```python
# all code is
!pip install --upgrade pip
%pip install wheel
%pip install datasets
%pip install --upgrade simplet5
from datasets import load_dataset

train_dataset = load_dataset("Xenova/sponsorblock-768", split="train")
test_dataset = load_dataset("Xenova/sponsorblock-768", split="test")
# make dataframe
train_df = train_dataset.to_pandas()
eval_df = test_dataset.to_pandas()
# extract text and extracted
train_df = train_df[["text", "extracted"]]
eval_df = eval_df[["text", "extracted"]]
# get first 1000 rows
train_df = train_df[:1000]
eval_df = eval_df[:100]
# rename columns
train_df.columns = ["source_text", "target_text"]
eval_df.columns = ["source_text", "target_text"]
# import
from simplet5 import SimpleT5

# instantiate
model = SimpleT5()
# load (supports t5, mt5, byT5 models)
model.from_pretrained("t5", "t5-small")
# train
model.train(
    train_df=train_df,  # pandas dataframe with 2 columns: source_text & target_text
    eval_df=eval_df,  # pandas dataframe with 2 columns: source_text & target_text
    source_max_token_len=512,
    target_max_token_len=128,
    batch_size=8,
    max_epochs=5,
    use_gpu=True,
    outputdir="outputs",
    early_stopping_patience_epochs=0,
    precision=32,
)

# load trained T5 model
model.load_model("t5", "path/to/trained/model/directory", use_gpu=False)

# predict
model.predict("input text for prediction")
```
